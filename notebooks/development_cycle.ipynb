```python
import pandas as pd

#loading the raw dataset
df=pd.read_csv("../data/raw/student_performance_dataset.csv")

print("Dataset Shape: ",df.shape)
print("\nMissing Values : ")
print(df.isnull().sum())

#stats
print("\nNumerical Summary:")
display(df.describe())

#distribution of results
print("\n Pass/Fail Counts : ")
print(df["Pass_Fail"].value_counts())

print("First Five Records :")
display(df.head())
```

```python
'''inspecting missing values'''

print("Attendence outside [0,100] :")
display(df[
    (df["Attendance_Rate"] < 0) |
    (df["Attendance_Rate"] > 100)
])

print("Past Exam Scores outside [0,100]:")
display(df[(df["Past_Exam_Scores"] < 0) | (df["Past_Exam_Scores"]>100)])

print("Final Exam Scores outside [0,100]:")
display(df[(df["Final_Exam_Score"]<0)| (df["Final_Exam_Score"]>100)])
```

```python
print("Duplicate student IDs:", df["Student_ID"].duplicated().sum())
print("Total missing values:", df.isnull().sum().sum())
```

```python
# Display duplicate student records

duplicates = df[df["Student_ID"].duplicated(keep=False)]

print("Total duplicate records :",len(duplicates))
display(duplicates.sort_values("Student_ID").head(20))
 
 #Number of unique students
print("Total records :",len(df))
print("Unique student IDs", df["Student_ID"].nunique())
```

```python
print("Invalid study hours:")
display(df[(df["Study_Hours_per_Week"]<0) | (df["Study_Hours_per_Week"]>60)])

print("Invalid final exam scores:")
display(df[(df["Final_Exam_Score"]<0) | (df["Final_Exam_Score"])>100])
```

```python
duplicates = df[df["Student_ID"].duplicated(keep=False)]
display(duplicates.sort_values("Student_ID"))
```

```python
# Check exact duplicate rows
print("Exact duplicate rows:",df.duplicated().sum())
# Check whether duplicate IDs have identical records
duplicate_groups = duplicates.groupby("Student_ID")
for student_id, group in duplicate_groups:
    if len(group.drop_duplicates()) == 1:
        print(student_id, "-> Identical records")
    else:
        print(student_id, "-> Different records")
```

```python
# Remove only exact duplicate rows
clean_df = df.drop_duplicates().copy()
print("Original records:", len(df))
print("Records after removing exact duplicates:", len(clean_df))
print("Rows removed:", len(df) - len(clean_df))
print("Remaining duplicate IDs :", clean_df["Student_ID"].duplicated().sum())
```

```python
clean_df.to_csv("../data/processed/student_analytics_cleaned.csv",index=False)

print("Cleaned dataset saved successfully")

```
